```python
import re
import pandas as pd

MAP = {1: 2, 2: 1, 3: 0, 4: -1, 5: 5}

def _apply_map(series: pd.Series) -> pd.Series:
    # Map only the specified values; keep others (e.g., 0, -1, NaN) unchanged
    return series.replace(MAP)

def _clean_var_name(rest: str) -> str:
    # Drop a leading underscore if present (handles 'Index_' vs 'Index')
    return rest[1:] if rest.startswith('_') else rest

def add_limonadi_columns(df: pd.DataFrame) -> pd.DataFrame:
    # Work on a copy if you want to keep original intact
    df = df.copy()

    # We’ll collect planned insertions first (to avoid shifting indices while iterating)
    insertions = []

    for col in list(df.columns):
        # Match Index* columns (with or without underscore)
        m_idx = re.match(r'^Index(_?.+)$', col)
        if m_idx:
            var = _clean_var_name(m_idx.group(1))
            new_col = f'Limonadi_Index_{var}'
            insertions.append(('index', col, new_col))
            continue

        # Match Overall* columns (with or without underscore)
        m_ov = re.match(r'^Overall(_?.+)$', col)
        if m_ov:
            var = _clean_var_name(m_ov.group(1))
            new_col = f'Limonadi_Overall_{var}'
            insertions.append(('overall', col, new_col))

    # Perform insertions in left-to-right order based on current column positions
    # (we compute fresh positions each time because the frame grows as we insert)
    for kind, src_col, new_col in insertions:
        # Build the new series
        new_series = _apply_map(df[src_col])

        # If the column already exists, update it and move it right after the source
        if new_col in df.columns:
            df[new_col] = new_series
            s = df.pop(new_col)
            pos = df.columns.get_loc(src_col) + 1
            df.insert(pos, new_col, s)
        else:
            pos = df.columns.get_loc(src_col) + 1
            df.insert(pos, new_col, new_series)

    return df

# --- Run it ---
combined_all = add_limonadi_columns(combined_all)

```
