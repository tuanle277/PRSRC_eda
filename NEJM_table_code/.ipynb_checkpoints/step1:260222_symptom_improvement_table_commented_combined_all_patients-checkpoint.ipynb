# --- Step 1: Identify Pre-op Symptoms ---

```python
def check_symptom_presence(row):
    presence = {}
    
    presence["Headache"] = row.get("Headache") == 1
    presence["DoubleVision"] = row.get("DoubleVision") == 1
    presence["SensorySymptoms"] = any([
        row.get("FacialNumbness") == 1,
        row.get("UpperExtremityNumbness") in [1,2,3],
        row.get("LowerExtremityNumbness") in [1,2,3],
        row.get("NeuropathicItch") == 1
    ])
    presence["WeaknessSymptoms"] = any([
        row.get("FacialWeakness") == 1,
        row.get("UpperExtremityWeakness") in [1,2,3],
        row.get("LowerExtremityWeakness") in [1,2,3]
    ])
    presence["Dysphagia"] = row.get("Dysphagia") == 1
    presence["Hoarseness"] = row.get("Hoarseness") == 1
    presence["Choking"] = row.get("Choking") == 1
    presence["SleepApnea"] = row.get("SleepApnea") == 1
    presence["ShortnessOfBreath"] = row.get("ShortnessOfBreath") == 1
    presence["NeckPain"] = row.get("NeckPain") == 1
    presence["BackPain"] = row.get("BackPain") == 1
    presence["TrunkPain"] = row.get("TrunkPain") == 1
    presence["UpperExtremityPain"] = row.get("UpperExtremityPain") in [1,2,3]
    presence["LowerExtremityPain"] = row.get("LowerExtremityPain") in [1,2,3]
    presence["Tremors"] = row.get("Tremors") == 1
    presence["BalanceOrGaitAtaxia"] = row.get("BalanceOrGaitAtaxia") == 1
    presence["Incontinence"] = row.get("Incontinence") == 1
    presence["SexualDysfunction"] = row.get("SexualDysfunction") == 1

    return pd.Series(presence)

preop_symptoms = symptom_history[['SubjectId']].copy()
symptom_presence = symptom_history.apply(check_symptom_presence, axis=1)
preop_symptoms = pd.concat([preop_symptoms, symptom_presence], axis=1)
```

# --- Step 2: Map follow-up symptoms ---

```python
symptom_map = {
    "Headache": "PreviouslyReportedHeadaches",
    "DoubleVision": "IndexDoubleVisionSymptoms",
    "SensorySymptoms": "IndexSensoryDeficitSymptoms",
    "WeaknessSymptoms": "IndexWeaknessSymptoms",
    "Dysphagia": "IndexDysphagiaSymptoms",
    "Hoarseness": "IndexHoarsenessSymptoms",
    "Choking": "IndexChokingSymptoms",
    "SleepApnea": "IndexSleepApneaSymptoms",
    "ShortnessOfBreath": "IndexSOBSymptoms",
    "NeckPain": "IndexNeckPainSymptoms",
    "BackPain": "IndexBackPainSymptoms",
    "TrunkPain": "IndexTrunkPainSymptoms",
    "UpperExtremityPain": "IndexUpperExtremPainSymptoms",
    "LowerExtremityPain": "IndexLowerExtremPainSymptoms",
    "Tremors": "IndexTremorsSymptoms",
    "BalanceOrGaitAtaxia": "IndexGaitAtaxiaSymptoms",
    "Incontinence": "IndexIncontinenceSymptoms",
    "SexualDysfunction": "IndexSexualDysfunctionSymptoms"
}


# Make sure treatment date is datetime format
treatment['ChiariSurgeryDate'] = pd.to_datetime(treatment['ChiariSurgeryDate'])

# Merge treatment date into followup
followup = followup.merge(treatment[['SubjectId', 'ChiariSurgeryDate']], on='SubjectId', how='outer')

followup['FollowupDate'] = pd.to_datetime(followup['FollowupDate'])
followup['DaysPostOp'] = (followup['FollowupDate'] - followup['ChiariSurgeryDate']).dt.days

# Filter to desired range
filtered_followup = followup[(followup['DaysPostOp'] >= 274) & (followup['DaysPostOp'] <= 729)]
```

```python
# Define helper function to check if a value is "unknown"
def is_unknown(symptom_key, value):
    return value == 5 or value == 0 or value == None

# Sort follow-ups
filtered_followup = filtered_followup.sort_values(['SubjectId', 'FollowupDate'])
filtered_followup['PreviouslyReportedHeadaches'] = filtered_followup['PreviouslyReportedHeadaches'].replace(0,5)
filtered_followup['IndexDoubleVisionSymptoms'] = filtered_followup['IndexDoubleVisionSymptoms'].replace(0,5)

# Initialize results
result_rows = []

# Loop by subject
for subject_id, group in filtered_followup.groupby('SubjectId'):
    result_row = {"SubjectId": subject_id}
    
    for preop_symptom, followup_symptom in symptom_map.items():
        symptom_series = group[followup_symptom]
        
        # if there is no records at all
        if not pd.notna(symptom_series).sum():
            result_row[followup_symptom] = 5
        else:
            # Reverse list of non-NaN values as default sorting is from small to large, early to late
            values = symptom_series.dropna().tolist()[::-1]  # latest first
            last_valid = None
            for val in values:
                # if we find any not "Unknown values"
                if not is_unknown(followup_symptom, val):
                    last_valid = val
                    break
            # if we fail to do so, it means we encounter 5 in the cords
            result_row[followup_symptom] = last_valid if last_valid is not None else 5
    
    result_rows.append(result_row)
# Convert to DataFrame
latest_followup_df = pd.DataFrame(result_rows)

# merge with site
if 'Site' in followup.columns:
    site_map = followup[['SubjectId', 'Site']].drop_duplicates()
    latest_followup_df = latest_followup_df.merge(site_map, on='SubjectId', how='outer')
latest_followup_df = latest_followup_df.fillna(5)
latest_followup_df
```

```python
# Merge pre-op and post-op data
merged = pd.merge(preop_symptoms, latest_followup_df, on='SubjectId', how='outer')

# Filter only patients who had any True in pre-op symptoms
symptom_cols = list(symptom_map.keys())
symptom_cols = [f"Preop_{col}" for col in symptom_cols]
merged['AnyPreopSymptom'] = merged[symptom_cols].any(axis=1)

#baseline_subjects_df = merged[merged['AnyPreopSymptom'] == True].drop(columns=['AnyPreopSymptom'])
baseline_subjects_df = merged.drop(columns=['AnyPreopSymptom'])
# Final per-subject output
subject_results_df = baseline_subjects_df.copy()
```

```python
# Explicit mapping between Preop and Index features
encoding_groups_oneb = {
    "DoubleVisionSymptoms": ("Preop_DoubleVision", "IndexDoubleVisionSymptoms"),
    "DysphagiaSymptoms": ("Preop_Dysphagia", "IndexDysphagiaSymptoms"),
    "HoarsenessSymptoms": ("Preop_Hoarseness", "IndexHoarsenessSymptoms"),
    "ChokingSymptoms": ("Preop_Choking", "IndexChokingSymptoms"),
    "SleepApneaSymptoms":("Preop_SleepApnea","IndexSleepApneaSymptoms"),
    "SOBSymptoms":("Preop_ShortnessOfBreath","IndexSOBSymptoms"),
    "NeckPainSymptoms":("Preop_NeckPain","IndexNeckPainSymptoms"),
    "BackPainSymptoms":("Preop_BackPain","IndexBackPainSymptoms"),
    "TrunkPainSymptoms":("Preop_TrunkPain","IndexTrunkPainSymptoms"),
    "UpperExtremityPainSymptoms":("Preop_UpperExtremityPain","IndexUpperExtremPainSymptoms"),
    "LowerExtremityPainSymptoms":("Preop_LowerExtremityPain","IndexLowerExtremPainSymptoms"),
    "TremorsSymptoms":("Preop_Tremors","IndexTremorsSymptoms"),
    "GaitAtaxiaSymptoms":("Preop_BalanceOrGaitAtaxia","IndexGaitAtaxiaSymptoms"),
    "IncontinenceSymptoms":("Preop_Incontinence","IndexIncontinenceSymptoms"),
    "SexualDysfunctionSymptoms":("Preop_SexualDysfunction","IndexSexualDysfunctionSymptoms")
}

def custom_index_encoding_oneb(preop, index):
    if preop == 0 and index == 1:
        return 1  
    elif preop == 0 and index == 2:
        return 2  
    elif preop == 0 and index == 3:
        return 3
    elif preop == 0 and index == 4:
        return 4
    elif preop == 0 and index == 0:
        return 3
    elif preop == 0 and index == 5:
        return 5
    elif preop == 0 and index == 6:
        return 6
    elif preop == 0 and index is None:
        return 5
    elif preop == 1 and index == 1:
        return 1  
    elif preop == 1 and index == 2:
        return 2  
    elif preop == 1 and index == 3:
        return 3  
    elif preop == 1 and index == 4:
        return 4  
    elif preop == 1 and index == 0:
        return 5
    elif preop == 1 and index == 5:
        return 5
    elif preop == 1 and index == 6:
        return 4
    elif preop == 1 and index is None:
        return 5
    return None  # fallback

# Apply to each mapping
for feature, (preop_col, index_col) in encoding_groups_oneb.items():
    subject_results_df[f'Index_{feature}'] = subject_results_df.apply(
        lambda row: custom_index_encoding_oneb(row[preop_col], row[index_col]), axis=1
    )
for feature, (preop_col, index_col) in encoding_groups_oneb.items():
    print(f"Processing feature: {feature}")
    print(subject_results_df[index_col].value_counts())
```

```python
import pandas as pd
import numpy as np

def extract_new_reported_symptoms(
    df_followup: pd.DataFrame,
    mapping: dict,                       # {preop_key: (index_col, new_col or None)}
    subject_id_col: str = "SubjectId",
    date_col: str = "FollowupDate",
) -> pd.DataFrame:
    """
    For each pre-op key in mapping, look up its 'new/overall' column (if any),
    and for each subject take the LATEST value that is NOT NaN, 0, or 5.
    If none found, set 5 (unknown).
    Output columns are named New_<preop_key>.
    """
    # sort ascending by date so we can scan from the end (latest first)
    df_sorted = df_followup.sort_values([subject_id_col, date_col], ascending=[True, True])

    def is_unknown(val):
        # Treat NaN, 0, 5 as "unknown/ignore"
        if pd.isna(val):
            return True
        try:
            v = float(val)
            return v in (5.0)
        except Exception:
            # non-numeric strings are considered valid (keep) unless NaN
            return False

    out_rows = []
    for sid, grp in df_sorted.groupby(subject_id_col, sort=False):
        row = {subject_id_col: sid}

        for preop_key, (_, new_col) in mapping.items():
            out_col = new_col.replace("_", "")
            # out_col = f"Overall_{preop_key}"

            # If no mapped new/overall column, mark unknown (5)
            if (out_col is None) or (out_col not in grp.columns):
                print(f"Subject {sid}, no column for {out_col}, setting to 5")
                row[new_col] = 5
                continue

            ser = grp[out_col]
            print(f"Subject {sid}, processing {out_col}, values: {ser.tolist()}")

            # Scan from latest to earliest, pick first non-unknown
            last_valid = 5
            # use .iloc[::-1] to iterate from latest
            for val in ser.iloc[::-1]:
                if not is_unknown(val):
                    last_valid = val
                    break

            row[new_col] = last_valid

        out_rows.append(row)

    return pd.DataFrame(out_rows)

new_symptom_flags = extract_new_reported_symptoms(filtered_followup, preop_to_post)
```

```python
def add_refined_symptoms(df: pd.DataFrame,
                         mapping: dict,
                         treat_zero_as_unknown: bool = True) -> pd.DataFrame:
    """
    For each mapping item {preop_key: (index_col, overall_col)}, create Refined_<IndexBase>:
      1) Refined = highest of (Index, Overall) ignoring 5 (Unknown) and NaN
      2) If both are Unknown/NaN → 5
      3) Any value 6 (New) is treated as 4 (Worse) before comparison
    """
    def proces(s: pd.Series) -> pd.Series:
        # Rule 3: 6 ("New") → 4 ("Worse")
        s = s.replace(6, 4)
        # mask unknowns
        mask = (s == 5) if treat_zero_as_unknown else False
        out = s.astype("float").copy()
        out[mask] = np.nan
        return out

    for _, (index_col, overall_col) in mapping.items():
        # derive base name from Overall_*
        print(index_col, overall_col)
        base = index_col.replace("Index", "")
        refined_col = f"Refined_{base}"

        idx = df[index_col] if index_col in df.columns else pd.Series(np.nan, index=df.index)
        ovr = df[overall_col] if (overall_col and overall_col in df.columns) else pd.Series(np.nan, index=df.index)

        idx_n = proces(idx)
        ovr_n = proces(ovr)

        # max ignoring NaNs; if both NaN → NaN, then fill with 5
        refined = pd.DataFrame({"i": idx_n, "o": ovr_n}).max(axis=1, skipna=True).fillna(5)

        # keep integer dtype if possible
        try:
            refined = refined.astype("Int64")
        except Exception:
            pass

        df[refined_col] = refined

    return df

subject_results_df = add_refined_symptoms(subject_results_df, preop_to_post)
```

```python
def set_stable_refined_for_symptoms(df, ordered_symptom_cols):
    """
    For each symptom (grouped as Preop, Index, Overall, Refined in ordered_symptom_cols):
      if Preop == 0 and Index == 1(absent) and Overall != 6  -> Refined = 3
    Skip Headache (Preop_Headache).
    Only updates rows/cols that exist in df.
    """
    cols = ordered_symptom_cols[:]

    # start after "SubjectId", then process groups of 4: Preop, Index, Overall, Refined
    i = 1
    n = len(cols)
    while i + 3 < n:
        preop_col   = cols[i]
        index_col   = cols[i + 1]
        overall_col = cols[i + 2]
        refined_col = cols[i + 3]

        # stop if we reached meta
        if preop_col == "Site": 
            break

        # skip Headache
        if preop_col == "Preop_Headache":
            i += 4
            continue

        # ensure columns exist before applying
        missing_any = any(c not in df.columns for c in (preop_col, index_col, overall_col, refined_col))
        if not missing_any:
            mask = (df[preop_col] == 0) & (df[index_col] == 1) & (df[overall_col] != 6)
            df.loc[mask, refined_col] = 3

            mask = (df[preop_col] == 0) & (df[index_col] == 2) & (df[overall_col] != 6)
            df.loc[mask, refined_col] = 3

            mask = (df[preop_col] == 0) & (df[index_col] == 5) & (df[overall_col] != 6)
            df.loc[mask, refined_col] = 3

        i += 4

    return df

subject_results_df = set_stable_refined_for_symptoms(subject_results_df, ordered_symptom_cols)
```

```python
violations = {}

# Check each mapping
for preop_col, index_col in mapping.items():
    # Condition: Preop == 0 and Index == 5
    mask = (subject_results_df[preop_col] == 0) & (subject_results_df[index_col] == 5)
    violations[preop_col] = mask.any()  # True if any violation exists

# Show results
for preop_col, has_violation in violations.items():
    print(f"{preop_col} → Violation exists? {has_violation}")
```

```python
# convert to Limonadi Scoring
def map_to_limonadi(value):
    if value == 1:
        return 2
    elif value == 2:
        return 1
    elif value == 3:
        return 0
    elif value == 4:
        return -1
    elif value == 5:
        return 5
```

```python
# Apply map_to_limonadi to all columns starting with 'Refined_' and create new 'Limonadi_' columns
for col in subject_results_df.columns:
    if col.startswith('Refined_'):
        limonadi_col = 'Limonadi_' + col.replace('Refined_', '')
        subject_results_df[limonadi_col] = subject_results_df[col].apply(map_to_limonadi)
```
