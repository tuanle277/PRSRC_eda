```python
import pandas as pd
combined_df = pd.read_csv('combined_all_preop_index_new_refined_limonadi_251024.csv')
```

```python
preop_to_refined_symptom_map = {
    "Preop_Headache": "Refined_PreviouslyReportedHeadaches",
    "Preop_DoubleVision": "Refined_DoubleVisionSymptoms",
    "Preop_SensorySymptoms": "Refined_SensoryDeficitSymptoms",
    "Preop_WeaknessSymptoms": "Refined_WeaknessSymptoms",
    "Preop_Dysphagia": "Refined_DysphagiaSymptoms",
    "Preop_Hoarseness_y": "Refined_HoarsenessSymptoms",
    "Preop_Choking": "Refined_ChokingSymptoms",
    "Preop_SleepApnea": "Refined_SleepApneaSymptoms",
    "Preop_ShortnessOfBreath": "Refined_SOBSymptoms",
    "Preop_NeckPain": "Refined_NeckPainSymptoms",
    "Preop_BackPain": "Refined_BackPainSymptoms",
    "Preop_TrunkPain": "Refined_TrunkPainSymptoms",
    "Preop_UpperExtremityPain": "Refined_UpperExtremPainSymptoms",
    "Preop_LowerExtremityPain": "Refined_LowerExtremPainSymptoms",
    "Preop_Tremors": "Refined_TremorsSymptoms",
    "Preop_BalanceOrGaitAtaxia": "Refined_GaitAtaxiaSymptoms",
    "Preop_Incontinence": "Refined_IncontinenceSymptoms",
    "Preop_SexualDysfunction": "Refined_SexualDysfunctionSymptoms",
}
```

```python

pfd_df = combined_df[combined_df['Allocation'] == 'PFD'].copy()
pfdd_df = combined_df[combined_df['Allocation'] == 'PFDD'].copy()

pfd_df = pfd_df.dropna(subset=['SubjectId'])
pfdd_df = pfdd_df.dropna(subset=['SubjectId'])

print("PFD group:", pfd_df.shape)
print("PFDD group:", pfdd_df.shape)
```

```python
list(pfd_df)
```

```python
# Function to generate summary table for a given group
def summarize_symptoms(df, label='Group'):
    all_summaries = []

    for pre_symptom, post_symptom in preop_to_refined_symptom_map.items():
        # filter with preoperatove data
        subset = df[df[pre_symptom] == True]
        baseline_count = subset['SubjectId'].nunique()

        temp = subset[['SubjectId', post_symptom]].fillna(5)
        temp = temp.rename(columns={post_symptom: 'Outcome'})
        temp['Outcome'] = temp['Outcome']
        temp['Symptom'] = pre_symptom

        count_df = temp.groupby('Outcome')['SubjectId'].nunique().reset_index()
        count_df['Symptom'] = pre_symptom
        count_df = count_df.rename(columns={'SubjectId': 'N_Patients'})
        count_pivot = count_df.pivot(index='Symptom', columns='Outcome', values='N_Patients')
        count_pivot['PresentBaseline'] = baseline_count
        all_summaries.append(count_pivot)
    # 4. Rename outcome columns
    outcome_labels = {
        1: 'Resolved',
        2: 'Improved',
        3: 'Stable',
        4: 'Worse',
        5: 'Unknown',
        0: 'Not Applicable',
        'Nan in the datasheets':'Nan in the datasheets',
        '1': 'Resolved',
        '2': 'Improved',
        '3': 'Stable',
        '4': 'Worse',
        '5': 'Unknown',
        '0': 'Not Applicable',
    }

    # 5. Reorder columns (only include if present)
    desired_order = [
        'PresentBaseline', 'Resolved', 'Improved', 'Stable', 'Worse',
        'Unknown','Not Applicable'
    ]

    # Combine and rename
    summary_df = pd.concat(all_summaries).fillna(0).astype(int)
    print(list(summary_df.columns))



    summary_df = summary_df.rename(columns=outcome_labels)
    for col in desired_order:
        if col not in summary_df.columns:
            summary_df[col] = 0

    # Reorder
    summary_df = summary_df[[col for col in desired_order if col in summary_df.columns]]
    summary_df = summary_df.reset_index()
    summary_df['Group'] = label

    return summary_df


# Apply separately to each group
pfd_summary = summarize_symptoms(pfd_df, label='PFD')
pfdd_summary = summarize_symptoms(pfdd_df, label='PFDD')

combined_summary = pd.concat([pfd_summary, pfdd_summary], ignore_index=True)

pfd_summary
```

```python
# Define all outcome-related columns to sum
outcome_cols = [
    'Resolved',
    'Improved',
    'Stable',
    'Worse',
    'Unknown',
]

# Check for mismatches in PFD
pfd_summary['TotalCounted'] = pfd_summary[outcome_cols].sum(axis=1)
pfd_summary['equal'] = pfd_summary['TotalCounted'] == pfd_summary['PresentBaseline'] 


# Check for mismatches in PFDD
pfdd_summary['TotalCounted'] = pfdd_summary[outcome_cols].sum(axis=1)
pfdd_summary['equal'] = pfdd_summary['TotalCounted'] == pfdd_summary['PresentBaseline'] 
```

```python
desired_order = list(preop_to_refined_symptom_map.keys())

# Ensure all needed columns exist
required_columns = ['Symptom', 'PresentBaseline', 'Resolved', 'Improved', 'Stable', 'Worse', 'Unknown', 'Group', 'TotalCounted', 'equal']
for col in required_columns:
    if col not in pfd_summary.columns:
        pfd_summary[col] = 0

# Set index to 'Symptom' for easy reindexing
pfd_summary.set_index('Symptom', inplace=True)

# Reindex with desired order and fill all missing values with NaN (temporarily)
pfd_summary_ordered = pfd_summary.reindex(desired_order).reset_index()

# Fill missing values in specific columns first
pfd_summary_ordered['Group'] = pfd_summary_ordered['Group'].fillna('PFD')
pfd_summary_ordered['equal'] = pfd_summary_ordered['equal'].fillna(True)

# Fill remaining numeric columns with 0
pfd_summary_ordered.fillna(0, inplace=True)


int_cols = ['PresentBaseline', 'Resolved', 'Improved', 'Stable', 'Worse', 'Unknown', 'TotalCounted']
pfd_summary_ordered[int_cols] = pfd_summary_ordered[int_cols].astype(int)
```
